```python
from openai import OpenAI
```

Check that you're connected to the LAN by going to http://192.168.88.10:7777/models (it should return json).

If having trouble, try turning off your main WiFi and just use the Ethernet. 

```python
client = OpenAI(
    base_url="http://192.168.88.10:7777/v1", # "http://<Your api-server IP>:port"
    api_key = "no-thanks"
)
```

```python
test_prompt = "hello from the Feminist AI LAN party!" # write whatever you want!
```

```python
messages = [{
    'role': 'user',
    'content': test_prompt
}]
```

```python
completion = client.chat.completions.create(
    model="qwen35_35",
    messages=messages
)
```
